```python
import pandas as pd
import numpy as np

# Mercedes-Benz Greener Manufacturing Dataset
# Kaggle URL: https://www.kaggle.com/aditya1702/mercedes-benz-data-exploration/data

train_url = 'https://raw.githubusercontent.com/krishnaik06/Complete-Feature-Engineering/master/mercedesbenz.csv'
train_df = pd.read_csv(train_url)

print("Train dataset shape:", train_df.shape)
display(train_df.head())
```

```python
train_df.describe()
```

```python
# Calculate variance only on numeric columns to avoid TypeErrors with categorical features
train_df.var(numeric_only=True)
```

```python
(train_df.var(numeric_only=True) == 0)
```

```python
variance_with_zero = train_df.var(numeric_only=True)[train_df.var(numeric_only=True) == 0].index.values
variance_with_zero
```

```python
train_df = train_df.drop(variance_with_zero, axis=1)
print(train_df.shape)
```

```python
train_df = train_df.drop(['ID'], axis=1)
```
